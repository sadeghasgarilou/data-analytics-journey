```python
#set dataframe
import pandas as pd

df = pd.DataFrame({
    "customer": ["Ali", "Sara", "Reza", "Neda", "Hassan"],
    "age": [25, None, 31, 28, None],
    "city": ["Tehran", "Shiraz", None, "Tabriz", "Tehran"],
    "income": [1200, 1800, 1500, None, 2100]
})

df
```

```python
#By default, dropna() removes every row that contains at least one missing value.
clean_df = df.dropna()
clean_df
```

```python
df.shape
```

```python
clean_df.shape
```

```python
#df.dropna() = df.dropna(axis=0, how="any")
df.dropna(axis=0,how='any')
```

```python
#Remove columns containing any missing value
df.dropna(axis=1)
```

This removes any column containing at least one missing value

```python
#Keep rows where at least one value is present
df.dropna(axis=0, how='all')
```

!!! IMPORTANT : Removing a row because any value is missing can be too aggressive. Real datasets often have thousands of rows, and you don't want to throw away useful information unnecessarily.

```python
#Check only important columns
#Imagine you're analyzing sales. An order without a Sales value may be unusable for a revenue analysis, but a missing City might not matter
df.dropna(subset=['income'])
```

This removes rows where income is missing, while allowing missing values in other columns.

```python
#We can check multiple columns
df.dropna(subset=['age','income'])
```

This removes a row if age or income is missing. Both fields must be present for the row to remain.

```python
#Require a minimum number of non-missing values : This is particularly useful when a dataset has many columns.
df.dropna(thresh=3)
```

keep a row if it has at least 3 non-missing values.

```python
#combine thresh with axis
df.dropna(thresh=4,axis=1)
```

```python
df.dropna(inplace=True)
```

This modifies df directly instead of returning a separate cleaned DataFrame.

```python
#Practice
import pandas as pd

customers = pd.DataFrame({
    "customer": ["Ali", "Sara", "Reza", "Neda", "Hassan", "Maryam"],
    "age": [25, None, 31, 28, None, 35],
    "city": ["Tehran", "Shiraz", None, "Tabriz", "Tehran", None],
    "income": [1200, 1800, 1500, None, 2100, 2400]
})

print(customers)
```

```python
#Remove rows that contain at least one missing value.
customers.dropna(axis=0, how='any')
```

```python
#Remove only rows where every column is missing.
customers.dropna(axis=0,how='all')
```

```python
#Remove rows where income is missing, ignoring missing values in other columns.
customers.dropna(axis=0, subset=['income'])
```

```python
#Keep rows that have at least 3 non-missing values.
customers.dropna(axis=0, thresh=3)
```

```python
#Remove columns that have fewer than 5 non-missing values.
customers.dropna(axis=1, thresh=5)
```
