```python
import pandas as pd

data = pd.DataFrame({
    "name": ["Ali", "Sara", "Reza", "Neda"],
    "age": [25, 30, None, 28],
    "salary": [1200, None, 1500, 1800]
})

data
```

Here Sara's Salary and Reza's age are missing Value

NaN means: Not a Number / missing value

we should investigate missing data before deciding what to do with it.

```python
#detect missing values
# isna() and isnull() : They are effectively equivalent.
data.isna()
```

Each True means:
There is a missing value here.

```python
#Count Missing Values
#True is 1 and False is zero. so we can use agg functions to get information about NaN.
data.isna().sum()
```

Whenever you receive a new dataset, one of the first things you should do is: data.insna().sum()

```python
#Missing percentage
data.isna().mean()*100
```

here True is 1 and False is zero. so we multiple mean 100 to calculate percentage of NaN.

```python
#Finding Rows With Missing Values
data.loc[data['age'].isna()]
```

```python
#Find rows where salary is missing
data.loc[data['salary'].isna()]
```

```python
#Find rows with missing values anywhere
data.loc[data.isna().any(axis=1)]
```

.any(axis=1) --> means: Does this row contain at least one True

```python
#practice 1
#create a new dataframe
practice = pd.DataFrame({
    "customer": ["Ali", "Sara", "Reza", "Neda", "Hassan", "Maryam"],
    "age": [25, None, 31, 28, None, 35],
    "city": ["Tehran", "Shiraz", None, "Tabriz", "Tehran", None],
    "income": [1200, 1800, None, 1500, None, 2100]
})

practice
```

```python
#How many missing values are there in each column?
practice.isna().sum()
```

```python
#Which rows have at least one missing value?
practice.loc[practice.isna().any(axis=1)]
```

```python
#How many missing values are there in the entire DataFrame?
practice.isna().sum().sum()
```

```python
#Calculate the missing percentage for every column and sort it from highest to lowest.
average_result = practice.isna().mean() * 100
average_result.sort_values(ascending=False)
```

```python
#Find age whose income is missing.
practice.loc[practice['age'].isna()]
```
