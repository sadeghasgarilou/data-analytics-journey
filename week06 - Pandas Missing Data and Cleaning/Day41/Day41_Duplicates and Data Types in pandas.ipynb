```python
#set dataframe
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "order_id": ["1001", "1002", "1002", "1003", "1004"],
    "customer": ["Ali", "Sara", "Sara", "Reza", "Neda"],
    "sales": ["1200", "850", "850", "1,500", "900"],
    "order_date": [
        "2026-01-10",
        "2026-01-11",
        "2026-01-11",
        "2026-01-12",
        "2026-01-13"
    ]
})

print(df)
print(df.info())
print(df.dtypes)
```

```python
#Find and remove duplicates
#Detect duplicate rows with duplicated()
df.duplicated()
```

Pandas returns a Boolean Series: True for rows considered duplicates and False otherwise.

```python
#Inspect the duplicate rows:
df.loc[df.duplicated()]
```

```python
#Count them:
df.duplicated().sum()
```

```python
#Remove duplicate rows with drop_duplicates()
df_clean = df.drop_duplicates()
df_clean
```

This keeps the first occurrence and removes subsequent identical rows.

```python
#Find duplicates based on selected columns
df.loc[df.duplicated(subset=['order_id'])]
```

This checks duplicates using only order_id.

```python
#inspect every row belonging to a repeated order ID:
df.loc[df.duplicated(subset=['order_id'],keep=False)]
```

The keep=False argument marks all occurrences of duplicated values as True, including the first one.

```python
#Choose which duplicate to keep
#The keep argument controls what duplicated() marks and what drop_duplicates() removes.
#keep="first"	Keep the first occurrence (default)
#keep="last"	Keep the last occurrence
#keep=False	Mark or remove every occurrence in a duplicate group
df_clean = df.drop_duplicates(keep=False)
df_clean
```

```python
#Understand and convert data types
#Data types tell Pandas how to interpret the values in each column.
#Inspect data types
df.dtypes
```

```python
#Hands-on cleaning workflow
#set messy dataframe
messy_sales = pd.DataFrame({
    "order_id": ["1001", "1002", "1002", "1003", "1004"],
    "city": [" Tehran ", "shiraz", "shiraz", "TEHRAN", None],
    "sales": ["1,200", "850", "850", "unknown", "900"],
    "order_date": [
        "2026-01-10",
        "2026-01-11",
        "2026-01-11",
        "2026-01-12",
        "2026-01-13"
    ]
})
```

```python
messy_sales
```

```python
#Inspect before cleaning
messy_sales.info()
```

```python
messy_sales.dtypes
```

```python
messy_sales.isna().sum()
```

```python
messy_sales.duplicated().sum()
```

```python
#Standardize the city text
messy_sales['city'] = (
    messy_sales['city']
    .str.strip()
    .str.title()
)
messy_sales
```

```python
#Convert sales to numeric values
messy_sales['sales'] = messy_sales['sales'].str.replace(',','',regex=False)
messy_sales['sales'] = pd.to_numeric(messy_sales['sales'], errors='coerce')
messy_sales
```

```python
messy_sales.loc[messy_sales['sales'].isna()]
```

```python
#Convert dates
messy_sales['order_date'] = pd.to_datetime(messy_sales['order_date'], errors='coerce')
messy_sales.dtypes
```

```python
#Remove exact duplicate rows
clean_sales = messy_sales.drop_duplicates()
clean_sales
```

```python
#solve NaN value
clean_sales['sales'] = clean_sales['sales'].fillna(clean_sales['sales'].mean())
clean_sales['city'] = clean_sales['city'].fillna('Unknown')
clean_sales
```

```python
#Validate
clean_sales.dtypes
```

```python
clean_sales.isna().sum()
```
