```python
#set dataframe
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "customer": [
        " Ali ",
        "SARA",
        "reza",
        "Neda  ",
        None,
        " ali "
    ],
    "city": [
        "Tehran",
        "shiraz",
        " TEHRAN ",
        "Tabriz",
        "Shiraz",
        None
    ],
    "category": [
        "Electronics",
        "electronics",
        " ELECTRONICS ",
        "Accessories",
        "accessories",
        "Electronics"
    ]
})

df
```

```python
#inspect distinct value
df['city'].unique()
```

```python
df['city'].value_counts(dropna=False)
```

```python
#Normalize capitalization and spaces
#Pandas provides a powerful string accessor called .str. It lets you apply text operations to an entire Series without writing a loop.
df['city'] = df['city'].str.lower()
df
```

Lowercase is especially useful for comparisons and standardizing category labels.

```python
#Convert text to uppercase or title case
df['customer'] = df['customer'].str.upper()
df
```

```python
df['city'].str.title()
```

```python
#Remove extra spaces with .str.strip()
df['customer'] = df['customer'].str.strip()
df
```

```python
df['city'] = df['city'].str.strip()
df
```

```python
#What if there are multiple spaces inside a string?
names = pd.Series(["Ali   Reza", "Sara  Ahmadi"])
names.str.replace(r'\s+',' ',regex=True)
```

The pattern \s+ matches one or more whitespace characters, replacing each run with one space.

```python
#Combine operations
df['city'] = (
    df['city']
    .str.strip()
    .str.upper()
)
df
```

```python
#Replace and search text
#Imagine a category column contains several variations:
categories = pd.Series([
    "Electronics",
    "electronics",
    "ELECTRONICS",
    "Electronic",
    "Accessories"
])
```

```python
#First, standardize capitalization:
categories = (
    categories
    .str.strip()
    .str.lower()
)
categories
```

```python
#Now you have consistent capitalization, but "electronic" is still different from "electronics".
categories = categories.replace({
    'electronic':'electronics'
})
categories
```

```python
#==
categories = (
    categories
    .str.strip()
    .str.lower()
    .replace({
        'electronic':'electronics'
    })
)
categories
```

```python
#Replace part of a string with .str.replace()
product_ids = pd.Series(["PR-101", "PR-102", "PR-103"])
product_ids.str.replace('-','',regex=False)
```

The regex=False argument tells Pandas to treat the search string literally rather than as a regular expression.

```python
#Search text with .str.contains()
df.loc[df['city'].str.contains('teh',na=False,case=False)]
```

Useful parameters:
- case=False: ignore capitalization.
- na=False: treat missing values as False.
- regex=False: search for a literal substring rather than a regular expression.

```python
#Search prefixes and suffixes
df.loc[df['city'].str.startswith('S',na=False)]
```

```python
df.loc[df['city'].str.endswith('Z',na=False)]
```

```python
#Practice challenge
messy = pd.DataFrame({
    "name": [" ali ", "SARA", "Reza  ", None, " ali"],
    "city": ["TEHRAN", " shiraz ", "tehran ", None, "TABRIZ"],
    "category": [
        "Electronics",
        "electronics ",
        "ACCESSORIES",
        "Accessories",
        None
    ]
})
```

```python
messy
```

```python
#Clean the name column by removing surrounding spaces and applying title case.
messy['name'] = messy['name'].fillna('Unknown')
messy['name'] = (
    messy['name']
    .str.strip()
    .str.title()
)
messy
```

```python
#Clean city values by stripping spaces and applying title case.
messy['city'] = messy['city'].fillna(messy['city'].mode())
messy['city'] = (
    messy['city']
    .str.strip()
    .str.title()
)
messy
```

```python
#Convert category values to lowercase and strip surrounding spaces.
messy['category'] = (
    messy['category']
    .str.strip()
    .str.lower()
)
messy
```

```python
#Find all rows where city contains 'teh', ignoring case and safely handling missing values.
messy.loc[messy['city'].str.contains('teh',na=False,case=False)]
```

```python
#Print the unique category values after cleaning.
messy['category'].unique()
```
