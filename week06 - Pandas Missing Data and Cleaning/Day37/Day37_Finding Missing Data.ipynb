```python
#Set Dataset
import pandas as pd

df = pd.DataFrame({
    "customer": ["Ali", "Sara", "Reza", "Neda", "Hassan"],
    "age": [25, None, 31, None, 29],
    "city": ["Tehran", "Shiraz", None, "Tabriz", "Tehran"],
    "income": [1200, 1800, None, 1500, 2100]
})

df
```

```python
#find missing value
df.isna()
```

```python
#find present value
df.notna()
```

```python
#find not NaN in a column
df['age'].notna()
```

```python
#find customer whose age is missing
df.loc[df['age'].isna()]
```

```python
#Which customers have a known income?
df.loc[df['income'].notna()]
```

```python
#Find records with missing age OR income
df.loc[(df['age'].isna()) | (df['income'].isna())]
```

```python
#Find records with both age AND income present
df.loc[(df['age'].notna()) & (df['income'].notna())]
```

```python
#Count missing values per column
df.isna().sum()
```

```python
#Count present values per column
df.notna().sum()
```

This tells us how many usable, non-missing entries each column contains. Whether those entries are actually valid still depends on the data.

```python
#Count rows with at least one missing value
df.isna().any(axis=1).sum()
```

```python
#Count rows where every value is present
df.notna().all(axis=1).sum()
```
