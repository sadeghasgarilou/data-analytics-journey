```python
# enumerate()
# When you want both the index and value
products = ["Laptop", "Mouse", "Keyboard"]
# in basic version
for i in range(len(products)):
    print(f"{i} {products[i]}")
```

```python
# but we can use enumerate() to do this :
for index, value in enumerate(products):
    print(f"{index} {value}")
```

```python
#Starting from 1
#By default, counting starts at 0.
#You can change that:
for index, product in enumerate(products, start=1):
    print(index, product)
```

```python
#Exercise 4
#Show students name by using their index
students = ["Ali", "Sara", "Reza", "Maryam", "Hassan"]
for index, student in enumerate(students, start=1):
    print(f"{index} - {student}")
```

```python
#Exercise 5
#enumerate and condition
scores = [85, 92, 58, 76, 95]
#print score when it is greater and equal than 80
for index, score in enumerate(scores):
    if score >= 80:
        print(f"{index}. {score}")
```

```python
#zip()
#when we want process two list together
products = ["Laptop", "Mouse", "Keyboard"]
prices = [1200, 25, 75]
#basic version
for i in range(len(products)):
    print(f"{products[i]} - {prices[i]}")
```

```python
#but by using of zip we can do it easily
for product, price in zip(products, prices):
    print(f"{product} - {price}")
```

```python
#Exercise 7
names = ["Ali", "Sara", "Reza"]
scores = [85, 92, 78]
#show student name and their scores by using of zip 
for name, score in zip(names, scores):
    print(f"{name} = {score}")
```

```python
#Exercise 8
#zip 3 lists
#zip can work with more than 2 lists
names = ["Ali", "Sara", "Reza"]
ages = [22, 24, 21]
scores = [85, 92, 78]
#show student's name with their age and scores
for name, age, score in zip(names, ages, scores):
    print(f"{name} <-> {age} <-> {score}")
```

```python
#Exercise 9
#zip + condition
products = ["Laptop", "Mouse", "Keyboard", "Monitor"]
prices = [1200, 25, 75, 350]
#show products with price greater than 100
for produc, price in zip(products, prices):
    if price > 100:
        print(f"{product} -> {price}")
```

```python
#Exercise 10
#zip + list comprehension
#create a list containing product and prices
produc_prices = [f"{product}: {price}" for product, price in zip(products, prices)]
print(produc_prices)
#So we can use zip() and also enumerate() in list comprehension
```

```python
#Important point 
#Normal zip() stops when the shortest iterable runs out.
names = ["Ali", "Sara", "Reza", "Maryam"]
scores = [85, 92]
print(list(zip(names,scores)))
```

```python
#Enumerate + Zip
products = ["Laptop", "Mouse", "Keyboard", "Monitor"]
prices = [1200, 25, 75, 350]
for index, (product, price) in enumerate(zip(products, prices), start=1):
    print(f"{index}. {product} -> {price}")
```

```python
#Exercise 11 and 12
#show the list with condition
for index,(product, price) in enumerate(zip(products, prices), start=1):
    if price > 100:
        print(f"{index}. {product} -> ${price}")
```

```python
#Date Science Challenge
products = [
    "Laptop",
    "Mouse",
    "Keyboard",
    "Monitor",
    "Desk",
    "Chair"
]

prices = [
    1200,
    25,
    75,
    350,
    300,
    150
]

quantities = [
    2,
    5,
    3,
    2,
    1,
    4
]

#Basic report
for index, (product, price, quantity) in enumerate(zip(products, prices, quantities), start=1):
    print(f"{index}. {product} -> ${price} * {quantity}")
```

```python
#calculate high revenue products
for index, (product, price, quantity) in enumerate(zip(products, prices, quantities), start=1):
    if (price * quantity) > 500:
        print(f"{index}. {product} -> ${price * quantity}")
```

```python
#List comprehension
#create revenues list with zip
revenues = [price * quantity for price, quantity in zip(prices, quantities)]
print(revenues)
```
