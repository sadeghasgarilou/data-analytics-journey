```python
#Multiple Parameters and default parameter
#Exercise 2
def calculate_price(price, quantity=1, discount=0):
    return (price * quantity) - (price * quantity * discount)
#Call with and without optional parameter
print(calculate_price(200))
print(calculate_price(200,2))
print(calculate_price(100,5,0.2))
```

```python
#Returning Multiple Values
#a function that return multiple values after calculation for a list of numbers
def calculation(numbers):
    total = sum(numbers)
    average = total / len(numbers)
    highest = max(numbers)
    return total, average, highest
#our list of numbers
numbers = [10,11,23,41,5]
#it return a tuple of result that we return
print(calculation(numbers))
#we can assign every return value to variable
#review tuple and unpacking it
sm, ag, mx = calculation(numbers)
print(sm)
```

```python
#Exercise 3
#a function that return average, highest, and least score
def analyze_score(scores):
    return sum(scores)/len(scores), max(scores), min(scores)
#call
scores = [15,18.6,12,11,6,20,19]
print(analyze_score(scores))
```

```python
#Exercise 4
students = [
    {"name": "Ali", "score": 85},
    {"name": "Sara", "score": 92},
    {"name": "Reza", "score": 58},
    {"name": "Maryam", "score": 76},
    {"name": "Hassan", "score": 95},
]
#define a function return of scores
def get_scores(students):
    return [student['score'] for student in students]
#define a function return of passed students scores
def get_passed_scores(students):
    return [student['score'] for student in students if student['score']>=60]
#define a function return count of passed students 
def count_passed_students(students):
    return len([student['score'] for student in students if student['score']>=60])
#call
print(get_scores(students))
print(get_passed_scores(students))
print(count_passed_students(students))
```

```python
#Function Calling function = Composition
#Exercise 5
def get_scores(students):
    return [student['score'] for student in students]
    
def calculate_average(scores):
    return sum(scores)/len(scores)
    
def analyze_students(students):
    scores = get_scores(students)
    average = calculate_average(scores)
    return average

print(analyze_students(students))
```

```python
#Final Exercise
def get_scores(students):
    return [student['score'] for student in students]
def analyze_students(scores):
    return sum(scores)/len(scores), max(scores), min(scores)
def classify_score(score):
    if score >= 90:
        return 'Excellent'
    elif score > 80:
        return 'Good'
    elif score > 60:
        return 'Average'
    else:
        return 'Failed'
def create_report(students):
    return [f'{student['name']} : {classify_score(student['score'])}' for student in students]
#Call it
print(create_report(students))
```
