```python
def classify_score(score):
    if score>=90:
        status = 'Excellent'
    elif score>80:
        status = 'Good'
    elif score>60:
        status = 'Average'
    else:
        status = 'Failed'
    return status
#call and print result for different valuse
result = classify_score(92)
print(result)
```

```python
#Exercise 4
#Functions  + Lists + Dictionaries
#my students data
students = [
    {"name": "Ali", "score": 85},
    {"name": "Sara", "score": 92},
    {"name": "Reza", "score": 58},
    {"name": "Maryam", "score": 76},
    {"name": "Hassan", "score": 95},
]
#define function that return passed students
def get_passed_students(students):
    result = []
    for student in students:
        if student['score']>=60:
            result.append(student)
    return result
#call function and see the result
final_passed_list = get_passed_students(students)
print(final_passed_list)
```

```python
#Docstring as help of function
def least_differ(a, b, c):
    """
    Return the smallest difference between any two numbers among a, b, and c
    >>> leat_difference(1,5,-5)
    4
    """
    diff1 = abs(a - b)
    diff2 = abs(a - c)
    diff3 = abs(b - c)
    return min(diff1, diff2, diff3)
```

```python
#get help using docstring that I wrote
help(least_differ)
```

```python
#Functions Applied to Functions
#supply functions as arguments to other functions.
#Functions that operate on other functions are called "higher-order functions."
def mod_5(x):
    """Return the remainder of x after dividing by 5"""
    return x % 5

print(
    'Which number is biggest?',
    max(100, 51, 14),
    'Which number is the biggest modulo 5?',
    max(100, 51, 14, key=mod_5),
    sep='\n',
)
```
