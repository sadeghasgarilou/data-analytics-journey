```python
#set dataset
import pandas as pd

sales = pd.DataFrame({
    "order_id": range(1001, 1013),
    "product": [
        "Laptop", "Mouse", "Keyboard", "Laptop",
        "Monitor", "Mouse", "Keyboard", "Monitor",
        "Laptop", "Mouse", "Monitor", "Keyboard"
    ],
    "category": [
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories", "Electronics", "Accessories"
    ],
    "salesperson": [
        "Ali", "Sara", "Ali", "Reza",
        "Sara", "Ali", "Reza", "Sara",
        "Reza", "Sara", "Ali", "Neda"
    ],
    "region": [
        "North", "South", "West", "East",
        "North", "East", "South", "West",
        "North", "South", "East", "West"
    ],
    "quantity": [2, 10, 5, 1, 3, 15, 8, 2, 2, 12, 4, 6],
    "price": [1000, 25, 50, 1000, 400, 25, 50, 400, 1000, 25, 400, 50]
})

sales["revenue"] = sales["quantity"] * sales["price"]
```

```python
#Total revenue by product
sales.groupby('product')['revenue'].sum()
```

```python
#The result is a Pandas Series, with product names as the index.
#To turn it into a DataFrame:
sales.groupby('product',as_index=False)['revenue'].sum()
#as_index=False keeps product as a regular column.
```

```python
#Let’s show the products with the highest revenue first:
sales.groupby('product',as_index=False)['revenue'].sum().sort_values(by='revenue',ascending=False)
```

```python
#How many orders did each product receive?
sales.groupby('product',as_index=False)['order_id'].count()
```

```python
#Multiple calculations with agg()
#Often, one summary is not enough. You may want total revenue, average revenue, quantity sold, and number of orders together.
sales.groupby('product',as_index=False).agg({'revenue':['sum','mean'],'quantity':['sum'],'order_id':['count']})
```

```python
#Try it: revenue by salesperson
#Total revenue
#Average order revenue
#Total quantity sold
#Number of orders
sales.groupby('salesperson').agg({'revenue':['sum','mean'],'quantity':['sum'],'order_id':['count']})
```

```python
#Combine filtering + grouping + sorting
#In the North and East regions, how much revenue did each product generate? Show the highest-revenue product first.
sales.loc[sales['region'].isin(['North','East'])].groupby('product',as_index=False)['revenue'].sum().sort_values(by='revenue',ascending=False)
```

```python
#mini challenge
#Revenue by region: Group by region and calculate total revenue. Sort from highest to lowest.
sales.groupby('region',as_index=False)['revenue'].sum().sort_values(by='revenue',ascending=False)
```

```python
#Product performance: Group by product and calculate total revenue, average revenue, and total quantity.
sales.groupby('product',as_index=False).agg(
    total_revenue=('revenue','sum'),
    average_revenue=('revenue','mean'),
    total_quantity=('quantity','sum')
)
```

```python
#Analyst question: Filter to the South and West regions, then group by salesperson and calculate total revenue. Sort descending.
sales.loc[sales['region'].isin(['South','West'])].groupby('salesperson',as_index=False).agg(
    total_revenue=('revenue','sum')
).sort_values('total_revenue')
```

Filter → Group → Aggregate → Sort
