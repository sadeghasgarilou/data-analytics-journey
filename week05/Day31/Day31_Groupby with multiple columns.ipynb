```python
#Create Dataset
import pandas as pd

sales = pd.DataFrame({
    "order_id": range(1001, 1013),
    "product": [
        "Laptop", "Mouse", "Keyboard", "Laptop",
        "Monitor", "Mouse", "Keyboard", "Monitor",
        "Laptop", "Mouse", "Monitor", "Keyboard"
    ],
    "category": [
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories", "Electronics", "Accessories"
    ],
    "salesperson": [
        "Ali", "Sara", "Ali", "Reza",
        "Sara", "Ali", "Reza", "Sara",
        "Reza", "Sara", "Ali", "Neda"
    ],
    "region": [
        "North", "South", "West", "East",
        "North", "East", "South", "West",
        "North", "South", "East", "West"
    ],
    "quantity": [2, 10, 5, 1, 3, 15, 8, 2, 2, 12, 4, 6],
    "price": [1000, 25, 50, 1000, 400, 25, 50, 400, 1000, 25, 400, 50]
})

sales["revenue"] = sales["quantity"] * sales["price"]
```

```python
#With one column: How much revenue did each product generate?
sales.groupby("product")["revenue"].sum()
```

```python
#With two columns: How much revenue did each product generate within each region?
sales.groupby(['region','product'],as_index=False)['revenue'].sum()
```

```python
#Multiple aggregations
#For every region/product combination, show total revenue, total quantity, average order revenue, and number of orders.
sales.groupby(['region','product'],as_index=False).agg(
    total_revenue = ('revenue','sum'),
    total_quantity = ('quantity','sum'),
    average_revenue = ('revenue','mean'),
    number_od_order = ('order_id','count')
)
```

```python
#Sort it
sales.groupby(['region','product'],as_index=False).agg(
    total_revenue = ('revenue','sum'),
    total_quantity = ('quantity','sum'),
    average_revenue = ('revenue','mean'),
    number_od_order = ('order_id','count')
).sort_values(by='total_revenue',ascending=False)
```

```python
#Three-column groupby()
sales.groupby(
    ["region", "category", "product"], as_index=False
).agg(
    total_revenue = ('revenue','sum'),
    total_quantity = ('quantity','sum')
)
```

```python
#Filter + Multiple groupby()
#For the North and East regions, show the performance of each salesperson for each product.
sales.loc[sales['region'].isin(['North','East'])].groupby(['product','salesperson'],as_index=False).agg(
    total_revenue=('revenue','sum'),
    total_quantity=('quantity','sum')
)
```
