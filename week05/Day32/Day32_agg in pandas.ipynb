```python
#Create Dataset
import pandas as pd

sales = pd.DataFrame({
    "order_id": range(1001, 1013),
    "product": [
        "Laptop", "Mouse", "Keyboard", "Laptop",
        "Monitor", "Mouse", "Keyboard", "Monitor",
        "Laptop", "Mouse", "Monitor", "Keyboard"
    ],
    "category": [
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories", "Electronics", "Accessories"
    ],
    "salesperson": [
        "Ali", "Sara", "Ali", "Reza",
        "Sara", "Ali", "Reza", "Sara",
        "Reza", "Sara", "Ali", "Neda"
    ],
    "region": [
        "North", "South", "West", "East",
        "North", "East", "South", "West",
        "North", "South", "East", "West"
    ],
    "quantity": [2, 10, 5, 1, 3, 15, 8, 2, 2, 12, 4, 6],
    "price": [1000, 25, 50, 1000, 400, 25, 50, 400, 1000, 25, 400, 50]
})

sales["revenue"] = sales["quantity"] * sales["price"]

```

```python
#first example
sales['revenue'].agg(['sum','min','median','max'])
```

```python
#use agg with multiple columns
sales.agg({
    'revenue' : ['sum','min','max'],
    'quantity' : ['sum']
})
```

```python
#Named Aggregation
#WHEN WE USE GROUPBY IT CAN BE SO HELPFUL
sales.groupby('product', as_index=False).agg(
    total_revenue=("revenue", "sum"),
    average_revenue=("revenue", "mean"),
    maximum_revenue=("revenue", "max"),
    minimum_revenue=("revenue", "min"),
    total_quantity=("quantity", "sum"),
    number_of_orders=("order_id", "count")
)
```

```python
#agg() + multiple-column groupby()
#Give me a detailed revenue report for every region and product.
sales.groupby(['region','product'],as_index=False).agg(
    total_revenue = ("revenue","sum"),
    average_revenue = ("revenue","mean"),
    maximum_revenue = ("revenue","max"),
    minimum_revenue = ("revenue","min"),
    total_quantity = ("revenue","sum"),
    number_of_order = ("order_id","count")
)
```

```python
#Mini Analyst Challenge
#Salesperson report
salesperson_report = sales.groupby('salesperson').agg(
    total_revenue = ('revenue','sum'),
    average_order_value = ('revenue','mean'),
    maximum_order_value = ('revenue','max'),
    minimum_order_value = ('revenue','min'),
    total_quantity = ('quantity','sum'),
    number_of_orders = ('order_id','count')
).sort_values(by="total_revenue", ascending=False)
```

```python
salesperson_report
```

```python
#region_category_report
region_category_report = sales.groupby(['region','category'],as_index=False).agg(
    total_revenue = ('revenue','sum'),
    average_revenue = ('revenue','mean'),
    total_quantity = ('quantity','sum'),
    number_of_orders = ('order_id','count') 
).sort_values(by='total_revenue', ascending=False)
```
