```python
#Create Dataset
import pandas as pd

sales = pd.DataFrame({
    "order_id": range(1001, 1013),
    "product": [
        "Laptop", "Mouse", "Keyboard", "Laptop",
        "Monitor", "Mouse", "Keyboard", "Monitor",
        "Laptop", "Mouse", "Monitor", "Keyboard"
    ],
    "category": [
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories", "Electronics", "Accessories"
    ],
    "salesperson": [
        "Ali", "Sara", "Ali", "Reza",
        "Sara", "Ali", "Reza", "Sara",
        "Reza", "Sara", "Ali", "Neda"
    ],
    "region": [
        "North", "South", "West", "East",
        "North", "East", "South", "West",
        "North", "South", "East", "West"
    ],
    "quantity": [2, 10, 5, 1, 3, 15, 8, 2, 2, 12, 4, 6],
    "price": [1000, 25, 50, 1000, 400, 25, 50, 400, 1000, 25, 400, 50]
})

sales["revenue"] = sales["quantity"] * sales["price"]
```

```python
sales
```

```python
#Warm up
#Filter
#Find orders from the North region:
sales.loc[sales['region'] == 'North']
```

```python
#GroupBy
#Find total revenue by product:
sales.groupby('product')['revenue'].sum()
```

```python
#Sort
#Find the highest-revenue orders:
sales.sort_values(by='revenue', ascending=False).head(3)
```

```python
#Exercise 1
#For the North and East regions, which products generated the most revenue?
sales.loc[sales['region'].isin(['North','East'])].groupby('product', as_index=False)['revenue'].sum().sort_values(by='revenue', ascending=False).head(2)
```

```python
#Ecercise 2
#Among orders with quantity greater than 5, how much revenue did each salesperson generate?
sales.loc[sales['quantity'] > 5].groupby('salesperson').agg(
    total_revenue = ('revenue','sum'),
    number_of_orders = ('order_id','count')
).sort_values(by='total_revenue', ascending=False)
```

```python
#Exercise 3
#For Electronics products sold in North or East, show revenue by salesperson and product.
sales.loc[(sales['category'] == 'Electronics') & (sales['region'].isin(['North','East']))].groupby(['salesperson','product'],as_index=False).agg(
    total_revenue = ('revenue','sum')
)
```

```python
#Final Challenge
#Analyze our high-value orders. I want to know which salesperson is performing best in the North and East regions.
#Define a high-value order as: revenue >= 500
sales.loc[sales['region'].isin(['North', 'East'])].groupby('salesperson').agg(
    total_revenue = ('revenue','sum')
).sort_values(by='total_revenue', ascending=False)
```
