```python
#Create Dataset
import pandas as pd

sales = pd.DataFrame({
    "order_id": range(1001, 1013),
    "product": [
        "Laptop", "Mouse", "Keyboard", "Laptop",
        "Monitor", "Mouse", "Keyboard", "Monitor",
        "Laptop", "Mouse", "Monitor", "Keyboard"
    ],
    "category": [
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories", "Electronics", "Accessories"
    ],
    "salesperson": [
        "Ali", "Sara", "Ali", "Reza",
        "Sara", "Ali", "Reza", "Sara",
        "Reza", "Sara", "Ali", "Neda"
    ],
    "region": [
        "North", "South", "West", "East",
        "North", "East", "South", "West",
        "North", "South", "East", "West"
    ],
    "quantity": [2, 10, 5, 1, 3, 15, 8, 2, 2, 12, 4, 6],
    "price": [1000, 25, 50, 1000, 400, 25, 50, 400, 1000, 25, 400, 50]
})

sales["revenue"] = sales["quantity"] * sales["price"]
```

```python
#review aggregation
sales['revenue'].sum()
```

```python
sales['revenue'].mean()
```

```python
sales['revenue'].nunique()
```

```python
#What is the total revenue for each product?
sales.groupby('product')['revenue'].sum()
```

```python
#For every product, find:
#- total revenue
#- average revenue
#- minimum revenue
#- maximum revenue
#- total quantity

sales.groupby('product', as_index=False).agg(
    total_revenue=('revenue','sum'),
    average_revenue=('revenue','mean'),
    minimum_revenue=('revenue','min'),
    maximum_revenue=('revenue','max'),
    total_quantity=('quantity','sum')
)
#This is called MULTIPLE AGGREGATION
```

```python
sales.groupby('salesperson',as_index=False).agg(
    total_revenue = ('revenue','sum'),
    average_revenue = ('revenue','mean'),
    total_quantity = ('quantity','sum'),
    number_of_order = ('order_id','count')
)
```

```python
#Multi-column groupby()
#Create a separate group for every region/product combination.
sales.groupby(['region','product'],as_index=False).agg(
    total_revenue = ('revenue','sum'),
    total_quantity = ('quantity','sum')
)
```

```python
#Add sorting
sales.groupby(['region','product'],as_index=False).agg(
    total_revenue = ('revenue','sum'),
    total_quantity = ('quantity','sum')
).sort_values(by='total_revenue',ascending=False)
```

```python
#Filtering before groupby()
#For only the North and East regions, what is the performance of each salesperson?
sales.loc[sales['region'].isin(['North','East'])].groupby('salesperson',as_index=False).agg(total_revenue=('revenue','sum')).sort_values(by='total_revenue')
```
