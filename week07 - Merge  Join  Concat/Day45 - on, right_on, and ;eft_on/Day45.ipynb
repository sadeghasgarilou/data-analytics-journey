```python
#set dataframe
import pandas as pd
import numpy as np

customers = pd.DataFrame({
    "customer_id": [101, 102, 103],
    "name": ["Ali", "Sara", "Reza"],
    "city": ["Tehran", "Shiraz", "Isfahan"]
})

orders = pd.DataFrame({
    "order_id": [1001, 1002, 1003],
    "customer_id": [101, 102, 101],
    "amount": [500, 750, 300]
})

display(customers)
display(orders)
```

```python
#Use on when the matching column has the same name in both DataFrames.
merged = pd.merge(
    customers,
    orders,
    on='customer_id'
)
display(merged)
```

```python
#Using left_on and right_on
#Two systems store the same information using different column names.
customers2 = pd.DataFrame({
    "customer_id": [101, 102, 103],
    "customer_name": ["Ali", "Sara", "Reza"]
})
orders2 = pd.DataFrame({
    "order_id": [1001, 1002, 1003],
    "client_id": [101, 102, 101],
    "amount": [500, 750, 300]
})
display(customers2)
display(orders2)
```

```python
#Both columns represent the same information, but their names differ.
#Solution: left_on and right_on
result = pd.merge(
    customers2,
    orders2,
    left_on = 'customer_id',
    right_on = 'client_id',
    how = 'inner'
)
display(result)
```

```python
#Because both columns contain the same matching identifiers, you might not need both in your final output.
result = result.drop(columns='client_id')
display(result)
```

```python
#Working with multiple merge keys
#Sometimes a single key isn't sufficient to identify the correct match.
targets = pd.DataFrame({
    "product": ["Laptop", "Laptop", "Phone", "Phone"],
    "region": ["East", "West", "East", "West"],
    "target": [100, 150, 200, 250]
})
sales = pd.DataFrame({
    "product": ["Laptop", "Laptop", "Phone", "Phone"],
    "region": ["East", "West", "East", "West"],
    "actual_sales": [90, 170, 220, 230]
})
display(targets)
display(sales)
```

```python
#Merge using two keys
result = pd.merge(
    targets,
    sales,
    on=['product','region']
)
display(result)
```

```python
#Multiple keys with different names
left_df = pd.DataFrame({
    "product": ["Laptop", "Phone"],
    "region": ["East", "West"],
    "target": [100, 200]
})

right_df = pd.DataFrame({
    "item": ["Laptop", "Phone"],
    "sales_region": ["East", "West"],
    "actual": [90, 220]
})

display(left_df)
display(right_df)
```

```python
result = pd.merge(
    left_df,
    right_df,
    left_on = ['product','region'],
    right_on = ['item','sales_region']
)
display(result)
```

```python
result = result.drop(columns=['item','sales_region'])
display(result)
```

```python
#Index-based merging
#So far, we've merged using ordinary columns.
#But sometimes the key is stored in a DataFrame's index.
#Pandas provides left_index and right_index for this.
employees = pd.DataFrame({
    "name": ["Ali", "Sara", "Reza"]
}, index=[101, 102, 103])

salaries = pd.DataFrame({
    "salary": [3000, 4000, 3500]
}, index=[101, 102, 104])

display(employees)
display(salaries)
```

```python
result = pd.merge(
    employees,
    salaries,
    left_index=True,
    right_index=True,
    how='outer'
)
```

```python
display(result)
```

```python
#Create the datasets
customers = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105],
    "customer_name": [
        "Ali", "Sara", "Reza", "Mina", "Hassan"
    ],
    "city": [
        "Tehran", "Shiraz", "Isfahan", "Tabriz", "Mashhad"
    ]
})

transactions = pd.DataFrame({
    "transaction_id": [1, 2, 3, 4, 5, 6],
    "client_id": [101, 101, 102, 103, 103, 106],
    "amount": [500, 300, 750, 200, 400, 900]
})

display(customers)
display(transactions)
```

```python
#Merge using different key names
result = pd.merge(
    customers,
    transactions,
    left_on='customer_id',
    right_on='client_id',
    how='left'
)
display(result)
```

```python
#Calculate total spending
result = result.drop(columns='client_id')
result_total = result.groupby(['customer_id','customer_name','city']).agg(
    total_amount = ('amount','sum'),
    order_counts = ('transaction_id','count')
)
display(result_total)
```

```python
#Find customers with no transactions
result_total.loc[result_total['order_counts'] == 0]
```

```python
#Investigate unmatched transactions
audit = pd.merge(
    customers,
    transactions,
    left_on = 'customer_id',
    right_on = 'client_id',
    how = 'outer',
    indicator = True
)
display(audit)
```

```python
#Find unmatched transactions:
audit.loc[audit['_merge'] == 'right_only']
```
