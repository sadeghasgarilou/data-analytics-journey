```python
import pandas as pd
import numpy as np
#Load and audit the dataset
df_raw = pd.read_csv(
    "MessyFoodPre.csv"
)

# Work on a copy; preserve the original data
df = df_raw.copy()
# Inspect structure and types
print("Dataset shape:", df.shape)
```

```python
#Find missing values
missing_report = pd.DataFrame({
    'missing_count' : df.isna().sum(),
    'missing_average' : (df.isna().mean() * 100).round(2)
})
missing_report
```

```python
#Find exact duplicates
print(f'Exact Duplicated : {df.duplicated().sum()}')
```

```python
#Inspect the duplicates:
duplicated = df.loc[df.duplicated(keep=False)]
duplicated
```

```python
#Inspect numerical statistics
df.describe()
```

```python
#Clean text columns
df['staff_experience'] = (
    df['staff_experience']
    .str.strip()
    .str.lower()
)
df.head()
```

```python
#Inspect the original categories
print(df['waste_category'].value_counts(dropna=False))
```

```python
#Standardise the categories
df['waste_category'] = (
    df['waste_category']
    .str.strip()
    .str.lower()
)
print(df['waste_category'].value_counts(dropna=False))
```

```python
#Investigate missing values
print(df['staff_experience'].value_counts(dropna=False))
```

```python
#Compare missing and non-missing rows
print(f'Missinga :')
display(df.loc[df['staff_experience'].isna()].head(10))
print(f'Not Missinga : ')
display(df.loc[df['staff_experience'].notna()].head(10))
```

```python
#Decide how to handle missing experience
df['staff_experience'] = (
    df['staff_experience']
    .fillna('Unknown')
)
```
