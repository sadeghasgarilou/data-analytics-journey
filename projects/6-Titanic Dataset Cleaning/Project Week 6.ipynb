```python
#Create a missing-values report
missing_values_reports = pd.DataFrame({
    'missing_count' : df.isna().sum(),
    'missing_percent' : (df.isna().mean() * 100).round(2)
})
missing_values_reports = missing_values_reports.sort_values(by='missing_count', ascending=False)
missing_values_reports
```

```python
#Inspect all occurrences of repeated passengers IDs without assuming they're errors:
df.loc[df.duplicated(subset=['PassengerId'])]
```

```python
#Inspect text values and numeric columns
text_columns = ['Name','Sex', 'Ticket', 'Cabin', 'Embarked']
for col in text_columns:
    print(f'col : {col}')
    print(df[col].value_counts(dropna=False).head(10))
    print()
```
