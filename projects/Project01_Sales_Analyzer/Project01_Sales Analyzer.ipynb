```python
#Data Set
sales = [
    {
        "product": "Laptop",
        "category": "Electronics",
        "price": 1200,
        "quantity": 2,
        "customer": "Ali"
    },
    {
        "product": "Mouse",
        "category": "Electronics",
        "price": 25,
        "quantity": 5,
        "customer": "Sara"
    },
    {
        "product": "Keyboard",
        "category": "Electronics",
        "price": 75,
        "quantity": 3,
        "customer": "Reza"
    },
    {
        "product": "Monitor",
        "category": "Electronics",
        "price": 350,
        "quantity": 2,
        "customer": "Maryam"
    },
    {
        "product": "Desk",
        "category": "Furniture",
        "price": 300,
        "quantity": 1,
        "customer": "Ali"
    },
    {
        "product": "Chair",
        "category": "Furniture",
        "price": 150,
        "quantity": 4,
        "customer": "Sara"
    },
    {
        "product": "Notebook",
        "category": "Stationery",
        "price": 5,
        "quantity": 10,
        "customer": "Hassan"
    },
    {
        "product": "Pen",
        "category": "Stationery",
        "price": 2,
        "quantity": 20,
        "customer": "Reza"
    }
]
```

DATA EXPLORATION

```python
count_of_sales = len(sales)
print(f'Count of records : {count_of_sales}')
```

```python
print(sales[0])
```

```python
print(sales[0]['product'])
print(sales[0]['category'])
print(sales[0]['price'])
print(sales[0]['quantity'])
```

```python
products = [sale['product'] for sale in sales]
print(products)
```

```python
categories = set([sale['category'] for sale in sales])
print(categories)
```

Sale static

```python
total_revenue = 0
for sale in sales:
    revenue = sale['price'] * sale['quantity']
    total_revenue += revenue
    print(f'revenue of {sale['product']} is {revenue}')

print()
print(f'Total revenue is {total_revenue}')
```

```python
average_revenue = total_revenue / len(sales)
print(average_revenue)
```

```python
#Product Analysis
```

```python
most_expensive_title = ''
most_expensive_price = 0
for sale in sales:
    if sale['price'] > most_expensive_price:
        most_expensive_price = sale['price']
        most_expensive_title = sale['product']
print(f'most ecpensive product is {most_expensive_title} with {most_expensive_price}$')
```

```python
most_expensive_title_r = ''
most_expensive_price_r = 0
for sale in sales:
    revenue = sale['price'] * sale['quantity']
    if revenue > most_expensive_price_r:
        most_expensive_price_r = revenue
        most_expensive_title_r = sale['product']
print(f'most ecpensive product is {most_expensive_title_r} with {most_expensive_price_r}$')
```

```python
quantities = [sale['quantity'] for sale in sales]
print(f'Sum of product is {sum(quantities)}')
```

```python
customers = set([sale['customer'] for sale in sales])
print(customers)
```

```python
for sale in sales:
    if sale['customer'] == 'Ali':
        print(sale['product'])
    
```

Customer Analysis

```python
csale = {}
for sale in sales:
    revenue = sale['price'] * sale['quantity']
    if sale['customer'] in csale.keys():
        csale[sale['customer']] = csale[sale['customer']] + revenue
    else:
        csale[sale['customer']] = revenue    
print(csale)
            
```
