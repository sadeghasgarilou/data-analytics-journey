```python
#create average column
students['average'] = ((students['math'] + students['python'] + students['statistics']) / 3).round(2)
students.head(3)
```

```python
#Question 1
#Who scored 90 or higher in Python?
students.loc[students['python'] >= 90]
```

```python
#Question 2
#Who has an average >= 85?
students.loc[students['average'] >= 85]
```

```python
#Question 3
#Who has Math >= 85 AND Python >= 85?
students.loc[(students['math'] >= 85) & (students['python'] >= 85)]
```

```python
#Question 4 
#Show only: name, python, average for students whose average is at least 85.
students.loc[students['average'] >= 85, ['name', 'python', 'average']]
```

```python
#create ranking
ranking = students.sort_values(by='average', ascending = False)
```

```python
ranking
```

```python
#find top students
top_student_index = students['average'].idxmax()
students.loc[top_student_index]
```

```python
#Answer Real Analysis Questions
#Question 1 — What is the average performance of the whole class?
students['average'].mean()
```

```python
#Question 2 — Subject performance
#What's the average score for each subject?
print(students[['math','python','statistics']].mean())
```

```python
#Question 3 — Students needing attention
#Let's define a student as needing attention if they scored below 70 in any subject.
students.loc[(students['python'] < 70) | (students['math'] < 70) | (students['statistics'] < 70)]
```

```python
#export analysis
ranking.to_csv('result.csv', index=False)
```
