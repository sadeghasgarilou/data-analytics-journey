```python
#Regional Performance
#Which regions generate the most sales and profit?
df.groupby('Region', as_index=False).agg(
    total_sale = ('Sales','sum'),
    total_profit = ('Profit','sum'),
    average_sale = ('Sales','mean'),
    average_profit = ('Profit','mean'),
    number_of_order = ('Order ID','count')
).sort_values(by='total_profit',ascending=False)
```

```python
#Category Performance
#Which product categories are driving sales and profit?
df.groupby('Category', as_index=False).agg(
    total_sale = ('Sales','sum'),
    total_profit = ('Profit','sum'),
    average_discount = ('Discount','mean'),
    number_of_order = ('Order ID','count')
).sort_values(by='total_profit', ascending=False)
```

```python
#High-Value Orders
#Let's define a high-value order as: Sales >= 1500
#Among high-value orders, which categories generate the most sales and profit?
high_value_orders = df.loc[df['Sales'] >= 1500].groupby('Category').agg(
    total_sale = ('Sales','sum'),
    total_profit = ('Profit','sum')
).sort_values(by='total_sale', ascending=False)
high_value_orders
```

```python
#Final Analyst Challenge
#I want to know which sub-categories are performing particularly well in the West region. Show me the top-performing sub-categories based on profit.
df.loc[df['Region'] == 'West'].groupby('Sub Category')['Profit'].sum().head(1)
```
