```python
#set up dataset
import pandas as pd

sales = pd.DataFrame({
    "order_id": range(1001, 1013),
    "product": [
        "Laptop", "Mouse", "Keyboard", "Laptop",
        "Monitor", "Mouse", "Keyboard", "Monitor",
        "Laptop", "Mouse", "Monitor", "Keyboard"
    ],
    "category": [
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories", "Electronics", "Accessories"
    ],
    "salesperson": [
        "Ali", "Sara", "Ali", "Reza",
        "Sara", "Ali", "Reza", "Sara",
        "Reza", "Sara", "Ali", "Neda"
    ],
    "region": [
        "North", "South", "West", "East",
        "North", "East", "South", "West",
        "North", "South", "East", "West"
    ],
    "quantity": [2, 10, 5, 1, 3, 15, 8, 2, 2, 12, 4, 6],
    "price": [1000, 25, 50, 1000, 400, 25, 50, 400, 1000, 25, 400, 50]
})

sales["revenue"] = sales["quantity"] * sales["price"]

sales.head()
```

```python
#Filtering practice
#Show all orders where the quantity is greater than or equal to 5.
sales.loc[sales['quantity'] >= 5]
```

```python
#Show orders where: Category is "Electronics" and Revenue is greater than 1000
sales.loc[(sales['category'] == 'Electronics') & (sales['revenue'] > 1000)]
```

```python
#Show orders from the "North" or "East" regions.
sales.loc[sales['region'].isin(['North','East'])]
```

```python
#Show orders where:
#The salesperson is Ali or Sara
#Quantity is greater than 5
sales.loc[(sales['salesperson'].isin(['Ali','Sara'])) & (sales['quantity'] > 5)]
```

```python
#Sorting practice
#Sort all orders by revenue, from highest to lowest.
sales.sort_values(by='revenue',ascending=False)
```

```python
#Filter orders from the "Accessories" category, then sort by quantity from highest to lowest.
sales.loc[sales['category'] == 'Accessories'].sort_values(by='quantity',ascending=False)
```

```python
#Sort by:
#Category, ascending
#Revenue, descending
sales.sort_values(by=['category','revenue'],ascending=[True,False])
```

```python
#Aggregation practice
#Total revenue across all orders
sales['revenue'].sum()
```

```python
#Average revenue per order
sales['revenue'].mean()
```

```python
#Total quantity sold
sales['quantity'].sum()
```

```python
#Maximum order revenue
sales['revenue'].max()
```

```python
#Total revenue from Electronics
sales.loc[sales['category'] == 'Electronics','revenue'].sum()
```

```python
#Average quantity for orders in the South region
sales.loc[sales['region'] == 'South','quantity'].mean()
```

```python
#Total revenue from Ali’s orders
sales.loc[sales['salesperson'] == 'Ali','revenue'].sum()
```

```python
#For all orders, calculate the sum, mean, minimum, and maximum of revenue.
sales['revenue'].agg(['sum','mean','min','max'])
```

```python
analys_result = sales.loc[(sales['region'].isin(['North','East'])) & (sales['category'].isin(['Electronics','Monitor']))].sort_values(by='revenue',ascending=False)
analys_result
```

```python
analys_result.agg({'revenue':['sum','mean','count'],'order_id':'count'})
```

```python
#Day28 - Complete project
##Advanced Analysis
#calculate total revenue for each product
sales.groupby('product')['revenue'].sum().sort_values(ascending=False)
```

```python
#Extend the analysis to show both total revenue and total quantity for each product.
sales.groupby('product').agg({'quantity':'sum','revenue':'sum'})
```

```python
##Now analyze salespeople.
#Create a summary containing:
#Total revenue
#Average revenue per order
#Total quantity sold
#Number of orders
sales.groupby('salesperson').agg({'revenue':['sum','mean'],'quantity':['sum','count']})
```
