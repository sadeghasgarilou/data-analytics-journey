```python
#import numpy
import numpy as np

#raw data
students = np.array([
    "Ali",
    "Sara",
    "Reza",
    "Maryam",
    "Hassan",
    "Neda"
])

subjects = np.array([
    "Math",
    "Python",
    "Statistics",
    "English"
])
#Each row belongs to one student. Each column belongs to one subject.
scores = np.array([
    [85, 90, 78, 88],
    [92, 95, 89, 91],
    [70, 65, 72, 68],
    [96, 98, 94, 95],
    [60, 75, 64, 70],
    [88, 84, 90, 86]
])
```

```python
#Explore data
print(f"Shape is : {scores.shape}")
print(f"Dimension is : {scores.ndim}")
print(f"The Number of totall scores is : {scores.size}")
```

```python
#Calculate each student's average
students_average = np.mean(scores, axis=1)
for i,student in enumerate(students):    
    print(f"{i+1}. {student} : {students_average[i]}")
```

```python
#Find the top student
top_student = np.argmax(students_average)
print(f"The top student is : {students[top_student]} with average score : {np.max(students_average)}")
```

```python
#Calculate average of each subject
subjects_average = np.mean(scores,axis=0)
for i,subject in enumerate(subjects):
    print(f"{i+1}. {subject} : {subjects_average[i]}")
```

```python
#Find Best Performin Subject
top_subject = np.argmax(subjects_average)
print(f"The Best performing subject is : {subjects[top_subject]} with average score of {np.max(subjects_average)}")
```

```python
#Find the highest and lowest score
print(f"The highest score is : {np.max(scores)}")
print(f"The lowest score is : {np.min(scores)}")
```

```python
#Find that highest and lowes scores belong to which subject and which student
highest_position = np.unravel_index(np.argmax(scores),scores.shape)
#np.unravel_index with parameters of item and shape get the postition of that item in nD array
print(f"Highest score : {scores[highest_position[0],highest_position[1]]}")
print(f"Student name : {students[highest_position[0]]}")
print(f"Subject name : {subjects[highest_position[1]]}")

lowest_position = np.unravel_index(np.argmin(scores),scores.shape)
print(f"lowest score : {scores[lowest_position[0],lowest_position[1]]}")
print(f"Student name : {students[lowest_position[0]]}")
print(f"Subject name : {subjects[lowest_position[1]]}")
```

```python
#Find Students who passed
passed_student = students[students_average >= 70]
print(passed_student)
```

```python
#count passed students in each subject
passed_per_subject = np.sum(scores >=70, axis=0)
for i, subject in enumerate(subjects):
    print(f"The passed students in {subject} : {passed_per_subject[i]}")
```
