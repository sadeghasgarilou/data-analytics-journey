```python
#list of students
students = [
    {"name": "Ali", "score": 85},
    {"name": "Sara", "score": 92},
    {"name": "Reza", "score": 58},
    {"name": "Maryam", "score": 76},
    {"name": "Hassan", "score": 45},
]
#print status
for student in students:
    if student['score']>=90:
        status = 'Excellent'
    elif student['score']>=80:
        status = 'Good'
    elif student['score']>=60:
        status = 'Average'
    else:
        status = 'Failed'
    print(f'{student['name']}: {status}')
```

```python
passed_students = [student for student in students if student['score']>=60]
print(passed_students)
```

```python
scores = [student['score'] for student in students]
average = sum(scores)/len(scores)
print(f'Average is {sum(scores)/len(scores)}')
print(f'Highest is {max(scores)}')
print(f'Lowest is {min(scores)}')
```

```python
higher_than_average = [student for student in students if student['score']>average]
print(len(higher_than_average))
print(average)
```
