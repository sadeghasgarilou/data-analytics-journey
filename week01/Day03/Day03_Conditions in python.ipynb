```python
#Condition + Loop
students = [
    {"name": "Ali", "score": 85},
    {"name": "Sara", "score": 92},
    {"name": "Reza", "score": 78},
    {"name": "Maryam", "score": 95},
]
for student in students:
    if(student['score']>=85):
        print(f'{student['name']}: Passed')
    else:
        print(f'{student['name']}: Failed')
```

```python
#Final Data Science Challenge
for student in students:
    if student['score']>=90:
        status = 'Excellent'
    elif(student['score']>=80):
        status = 'Good'
    elif(student['score']>=70):
        status = 'Average'
    else:
        status = 'Poor'
    print(f'{student['name']}: {status}')
```
