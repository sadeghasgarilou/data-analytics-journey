List is combine of elemnt's in one or different types that we can access it's elements by indexing

```python
#Accessing Elements
sales = [100, 200, 150, 400]
print(sales[0])
print(sales[-1])
```

With Slicing we can access to a slice of list [start:stop:step]

```python
#Slicing
print(sales[0:2])
print(sales[:3])
print(sales[2:])
print(sales[::2])
```

List's are mutable and we can change it's values.

```python
#Mutable
data = ["Sadegh", 24, True, 23.21, "Done"]
print(data)
data[0] = "Asgarilou"
print(data)
```

There are different methode to work with lists. append for add at the end of list. insert for add data in the index that we want. remove for delete element. pop for delete an value by it's index. len for know lenght of list. .. in .. to know that an element exist in list or not. sort and sorted for sorting list, sort change original list but sorted return new list.

```python
#Methodes
data.append("Fati")
print(data)
data.insert(0,"Sadegh")
print(data)
data.remove(True)
print(data)
result = data.pop(3)
print(result)
print(data)
len(data)
print(24 in data)
numbers = [24,2,12,45,21]
numbers.sort(reverse=True)
print(numbers)
new_numbers = sorted(numbers)
print(new_numbers)
for number in numbers:
    print(f"The number is : {number+2}")
```

List Comperhension is important. [calculation for element in list]

```python
#List Comprehension
squares = []
for i in range(1,10):
    squares.append(i**2)

new_squares = [i**2 for i in range(1,10)]

print(squares)
print(new_squares)
```

Dictionary is a data type that hold value with it's key. Store data in key-value format.

```python
#Dictionaries
person = {
    "name" : "Sadegh",
    "age" : 31,
    "city" : "Zanjan"
}
```

Access value by it's key

```python
#Accessing
print(person["name"])
```

Add new value by it's key

```python
#Adding New Data
person['job'] = 'Software Developer'
print(person)
```

Change a value by it's key.

```python
#Changing Value
person["name"] = "Sadegh Asgarilou"
print(person)
```

remove by del+key or pop+key. pop return the value that is removed

```python
#Removing Values
del person["city"]
print(person)
job = person.pop('job')
print(job)
print(person)
```

With key in dictionary we can check that if the key exist in that dictionary or not

```python
#Checking Keys
print("name" in person)
```

.items() return key value set in tuple format

```python
#KEYS, VALUES, ITEMS
print(person.keys())
print(person.values())
print(person.items())
for key, value in person.items():
    print(f'The Value of {key} is {value}')
```

We can use get('key','default value') to not get error when that key don't exist

```python
#Get
#print(person["job"])
print(person.get("job","Unknown"))
```

```python
#List and Dictionaries
students = [
    {"name": "Ali", "age": 22, "score": 85},
    {"name": "Sara", "age": 24, "score": 92},
    {"name": "Reza", "age": 21, "score": 78}
]
print(students[0]["name"])
```

```python
for student in students:
    print(f'{student["name"]}\'s score is {student["score"]}')
```

```python
scores = [student['score'] for student in students]
average = sum(scores)/len(scores)
print(average)
```

```python
max_score = [student['name'] for student in students if student['score']==max(scores)]
print(max_score)
```
