```python
#Mixed Practice
students = [
    {
        "name": "Ali",
        "age": 22,
        "score": 85,
        "skills": ["Python", "SQL"]
    },
    {
        "name": "Sara",
        "age": 24,
        "score": 92,
        "skills": ["Python", "Pandas", "SQL"]
    },
    {
        "name": "Reza",
        "age": 21,
        "score": 58,
        "skills": ["Python"]
    },
    {
        "name": "Maryam",
        "age": 23,
        "score": 76,
        "skills": ["SQL", "Excel"]
    },
    {
        "name": "Hassan",
        "age": 25,
        "score": 95,
        "skills": ["Python", "Pandas", "NumPy"]
    }
]

students_name = [student['name'] for student in students]
print(students_name)
```

```python
students_score = [student['score'] for student in students]
print(students_score)
```

```python
average = sum(students_score)/len(students_score)
print(f'averge : {average}')
highest_score = max(students_score)
print(f'Highest : {highest_score}')
minest_score = min(students_score)
print(f'Minest : {minest_score}')
```

```python
for student in students:
    if student['score']>=90:
        status = 'Excellent'
    elif student['score']>80:
        status = 'Good'
    elif student['score']>60:
        status = 'Average'
    else:
        status = 'Failed'
    print(f'{student['name']} : {status}')
```

```python
passed_students = [student for student in students if student['score']>=60]
print(passed_students)
```

```python
excellent_students = [student for student in students if student['score']>=90]
print(excellent_students)
```

```python
max_score = 0
best_student = ''
for student in students:
    if student['score'] > max_score:
        max_score = student['score']
        best_student = student['name']
print(f'Best student is : {best_student} with score of {max_score}')
```

```python
skills = []
for student in students:
    for skill in student['skills']:
        skills.append(skill)
skills = set(skills)
print(skills)
```

```python
python_count = 0
for student in students:
    if 'Python' in student['skills']:
        python_count += 1
print(f'Number of students with python skills : {python_count}')
```

```python
expected = [student['name'] for student in students if student['age']>=23 and student['score']>=80]
print(expected)
```

```python
print('==== STUDENT REPORT ====')
print()
print(f'Total Student : {len(students)}')
print(f'Average score : {average}')
print(f'Highest score : {highest_score}')
print(f'Lowest score : {minest_score}')
print()
print('Passed Students')
for student in passed_students:
    print(student['name'])
print()
print('Excellent students')
for student in excellent_students:
    print(student['name'])
print()
print('Unique skills')
for skill in skills:
    print(skill)
print()
print(f'Python Users : {python_count}')
```
