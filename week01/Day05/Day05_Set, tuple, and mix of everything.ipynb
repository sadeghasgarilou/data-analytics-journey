```python
#list
numbers = [10, 20, 30]
#tuple
numbers = (10,20,30)
#tuple is immutable
numbers[0]=1
```

```python
#Accessing tuple elements
print(numbers[0])
```

```python
#tuple unpacking
person = ('Sadegh',31,'Python')
name, age, pl = person
print(name)
print(age)
print(pl)
```

```python
#To create tuple with one elemnt 
a = (10) #this is integer
b = (10,) #this create tuple
print(type(a))
print(type(b))
```

```python
#A set automatically removes duplicates.
n = {1,2,3,3,3,2,1,2,4}
print(n)
```

```python
#convert list to set
skills = [
    "Python",
    "SQL",
    "Python",
    "Pandas",
    "SQL",
    "Python"
]
unique_skills = set(skills)
print(unique_skills)
```

```python
#set does not use indexes
print(skills[0])
print(unique_skills[0])
```

```python
#Adding to and Removing from Set
skills = {'word', 'python'}
skills.add('Java')
print(skills)
skills.remove('word')
print(skills)
```

```python
#Set Operations
python_students = {"Ali", "Sara", "Reza"}
sql_students = {"Sara", "Reza", "Maryam"}
#Intersection
print(python_students & sql_students)
#Union
print(python_students | sql_students)
#Difference
print(python_students - sql_students)
```
