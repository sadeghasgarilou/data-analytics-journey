```python
#Exercise 1
students = [
    {"name": "Ali", "score": 85},
    {"name": "Sara", "score": 92},
    {"name": "Reza", "score": 58},
    {"name": "Maryam", "score": 76},
    {"name": "Hassan", "score": 45},
    {"name": "Nima", "score": 91},
]
total_score = 0
number_of_passed = 0
for student in students:
    total_score += student['score']
    if student['score']>=90:
        number_of_passed += 1
        status = 'Excellent'
    elif student['score']>80:
        number_of_passed += 1
        status = 'Good'
    elif student['score']>60:
        number_of_passed += 1
        status = 'Average'
    else:
        status = 'Failed'
    print(f'{student['name']} : {status}')
print(f'Total Score is : {total_score}')
print(f'Number of Passed Student is : {number_of_passed}')
```
