```python
#Review
students = [
    {"name": "Ali", "score": 85},
    {"name": "Sara", "score": 92},
    {"name": "Reza", "score": 58},
    {"name": "Maryam", "score": 76},
]
print(students[1]['score'])
all_scores = [student['score'] for student in students]
print(all_scores)
print(sum(all_scores)/len(all_scores))
passed_student = [student['name'] for student in students if student['score']>=60]
print(passed_student)
```

For is use to Iterate amount data. like list, dictionaries, and ...

```python
#For Loops
numbers = [1,5,8,21]
for number in numbers:
    print(f'{number} square = {number ** 2}')
```

range() give us a list of number based on the arguments that we've entered

```python
#range
#range(start, stop, step)
for i in range(1,10,2):
    print(i)
```

```python
for i in range(10,4,-1):
    print(i)
```

```python
#Loops + Condition
fruits = ['apple','melon','orange','watermelon','strawberry']
for fruit in fruits:
    if len(fruit)>5:
        print(fruit)
```

```python
numbers = [12, 5, 8, 21, 30, 7, 16]
for number in numbers:
    if number % 2 ==0:
        print(number)
```

```python
scores = [45, 82, 67, 91, 58, 76]
for score in scores:
    if score>=90:
        status='Excellent'
    elif score>80:
        status='Excellent'
    elif score>60:
        status='Average'
    else:
        status='Failed'
    print(f'{score} : {status}')
```

```python
#Loops + Dictionary
students = [
    {"name": "Ali", "score": 85},
    {"name": "Sara", "score": 92},
    {"name": "Reza", "score": 58},
    {"name": "Maryam", "score": 76},
]
total_score = 0
for student in students:
    total_score = total_score + student['score']
    print(f"{student['name']} -> {student['score']}")
print(f'total score is : {total_score}')
```

While Loop is another way of iterating amoung data

```python
#While loop
count = 1
while count<10:
    print(count)
    count +=1
```

```python
count = 10
while count>=1:
    print(count)
    count -= 1
```

```python
#Break
for number in range(10):
    if(number == 5):
        break
    print(number)
```

```python
numbers = [3, 7, 12, 18, 25, 30]
for number in numbers:
    if(number == 18):
        break
    print(number)
```

```python
for i in range(5):
    if(i==2):
        continue
    print(i)
```
