We can change variable amount and even type with assign it a new value

```python
age = 25
print(age)
age = 31
print(age)
```

```python
#Dynamic Typing
age = 25
print(type(age))
age = 'Twenty'
print(type(age))
```

We can do calculation in variables

```python
price = 10
count = 5
total_price = price * count
print(total_price)
```

```python
#CALCULATION
a = 10
b = 2
print(a + b)
print(a - b)
print(a / b)
#floor division with //
print(a // b)
print(a * b)
print(a ** b)
print(a % b)
```

Assignment shortcut is += -= *= and ...

```python
#Assignment shortcuts
x = 10 
# x = x +5
x += 5 
print(x)
```

We can combine strings with +

```python
#String
first_name = "Sadegh"
last_name = "Asgarilou"
full_name = first_name + " " + last_name
print(full_name)
```

f-string in modern python is a way to combine and create strings

```python
#f-string in modern python
age = 25
name = "Sadegh"
message = f"My name is {name} and I am {age} years old."
print(message)
```

```python
#Conversions
age = "25"
print(type(age))
age = int(age)
print(type(age))
```

This is a python dictionary. We will read about it in next days. It's a complicated type of variables in python that save data in a key-value format.

```python
profile = {
    "name" : "Sadegh",
    "age" : 31,
    "job" : "software developer",
    "goal" : "data analyst",
    "height" : 1.90,
    "is_student" : False
}
print(type(profile["is_student"]))
```
