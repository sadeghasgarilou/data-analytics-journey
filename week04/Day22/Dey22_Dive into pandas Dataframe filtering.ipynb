```python
#core concept
import pandas as pd

employees = pd.DataFrame({
    "name": [
        "Ali", "Sara", "Reza", "Maryam",
        "Hassan", "Neda", "Omid", "Zahra"
    ],
    "department": [
        "Data", "Web", "Data", "Backend",
        "Web", "Data", "Backend", "Web"
    ],
    "age": [25, 31, 28, 35, 24, 29, 32, 27],
    "salary": [2500, 3200, 2800, 4100, 2300, 3500, 3900, 3000],
    "experience": [2, 5, 3, 8, 1, 6, 7, 4],
    "performance": [82, 91, 76, 95, 68, 88, 93, 85]
})

employees
```

```python
#Boolean filtering
employees[employees['salary']>3000]
```

```python
#AND, OR, Not
#&, |, ~
#Find employees who: work in Data AND earn at least 3000
employees.loc[(employees['department']=='Data') & (employees['salary']>=3000)]
```

```python
# Find employees who: work in Data OR earn more than 4000
employees.loc[(employees['department']=='Data') | (employees['salary']>4000)]
```

```python
# Find everyone who is NOT in Data:
employees.loc[~(employees['department']=='Data')]
```

```python
#Practice 1
#1-Employees with: age >= 30 AND performance >= 90
employees.loc[(employees['age'] >= 30) & (employees['performance'] >= 90)]
```

```python
#2-Employees with: salary < 3000 OR performance < 75
employees.loc[(employees['salary'] < 3000) | (employees['performance'] < 75)]
```

```python
#3-Employees who are not in Backend.
employees.loc[~(employees['department']=='Backend')]
```

```python
#Professional Filtering with .loc
employees.loc[employees['age']>30,['name','age']]
```

```python
#isin
#Suppose you want employees from either: Data, Backend
employees.loc[(employees['department'] == 'Data') | (employees['department'] == 'Backend')]
```

```python
#Intead of code above we can use isin()
employees.loc[employees['department'].isin(['Data','Backend'])]
```

```python
#another example
employees.loc[employees['name'].isin(['Ali','Omid'])]
```

```python
#between
#Suppose we want employees between 25 and 30 years old.
employees.loc[(employees['age'] >= 25) & (employees['age'] <= 30)]
```

```python
#Intead of code above we can use between()
employees.loc[employees['age'].between(25,30)]
```

```python
#By default, between() includes both endpoints.
```

```python
#Names starting with "A"
employees.loc[employees['name'].str.startswith('A')]
```

```python
#Names ending with "a"
employees.loc[employees['name'].str.endswith('a')]
```

```python
employees.loc[employees['department'].str.contains('data', case=False, na=False)]
```

```python
#missing values
employees.loc[2, "salary"] = None
employees.loc[5, "performance"] = None
```

```python
#Find rows where salary is missing
employees.loc[employees['salary'].isna()]
```

```python
#Find rows where salary is NOT missing
employees.loc[employees['salary'].notna()]
```

```python
employees.query('salary > 3000')
```

```python
#multiple conditions
employees.query('salary > 3000 and age > 30')
```

```python
#age range
employees.query(' 25 < age < 30')
```

```python
#Filtering + Sorting + Selecting
#Find Data employees with performance >= 80, show their name, salary and performance, and sort them from highest performance to lowest.
employees.loc[(employees['performance'] >= 80) & (employees['department']=='Data'),['name','salary','performance']].sort_values(by='performance',ascending=False)
```

```python
#Find employees whose name contains "a" regardless of capitalization.
employees.loc[employees['name'].str.contains('a',case=False)]
```

```python
#Create a professional analysis:
#Find employees who are in the Data department, 
#have at least 3 years of experience, 
#and have performance >= 80. 
#Show only their name, salary, experience and performance, 
#then sort them by performance from highest to lowest.
employees.loc[(employees['department']=='Data') & (employees['experience'] >= 3),['name','salary','experience','performance']].sort_values('performance',ascending=False)
```
