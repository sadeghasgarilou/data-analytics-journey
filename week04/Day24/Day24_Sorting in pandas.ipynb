```python
#create our dataframe
import pandas as pd

employees = pd.DataFrame({
    "name": [
        "Ali", "Sara", "Reza", "Maryam",
        "Hassan", "Neda", "Omid", "Zahra"
    ],
    "department": [
        "Data", "Web", "Data", "Backend",
        "Web", "Data", "Backend", "Web"
    ],
    "age": [25, 31, 28, 35, 24, 29, 32, 27],
    "salary": [2500, 3200, 2800, 4100, 2300, 3500, 3900, 3000],
    "experience": [2, 5, 3, 8, 1, 6, 7, 4],
    "performance": [82, 91, 76, 95, 68, 88, 93, 85]
})

employees
```

```python
#basic sorting
employees.sort_values('age')
```

```python
#by default sort is ascending
#descending sort
employees.sort_values('age',ascending=False)
```

```python
#Practice 1
#Sort employees by performance, lowest to highest.
employees.sort_values('performance')
```

```python
#Sort employees by performance, highest to lowest.
employees.sort_values('performance',ascending=False)
```

```python
#Sorting Multiple Columns
employees.sort_values(['performance','salary'])
```

1. Sort by performance
2. If performance is equal,
   sort by salary

```python
#Performance: highest first, and Salary: lowest first
employees.sort_values(['performance','salary'],ascending=[False,True])
```

```python
#Practice 2
#department → A-Z
#salary → high to low
employees.sort_values(['department','salary'],ascending=[True,False])
```

```python
#experience → high to low
#performance → high to low
employees.sort_values(['experience','performance'],ascending=[False,False])
```

```python
#Sorting Strings
#This sorts alphabetically.
employees.sort_values('name')
```

```python
employees.sort_values(['department','name'])
```

```python
#Missing Values While Sorting
#Create missing values
employees.loc[2, "salary"] = None
employees.loc[5, "performance"] = None
```

```python
employees.sort_values('salary')
```

```python
#Pandas normally places missing values at the bottom.
#control NaN position in sort
employees.sort_values('salary',na_position='first')
```

```python
#Sorting Index with sort_index()
#sort based on dataframe index
employees.sort_index(ascending=False)
```

```python
#Sorting After Filtering
#Show Data employees and rank them by performance.
employees.loc[employees['department']=='Data'].sort_values('performance',ascending=False)
```

```python
#Top N / Bottom N
#Suppose you want the top 3 performers.
employees.sort_values('performance',ascending=False).head(3)
```

```python
#worst 3 performance
employees.sort_values('performance',ascending=False).tail(3)
```

```python
#nlargest()
employees.nlargest(3,'performance')
```

```python
#nsmallest
employees.nsmallest(3,'performance')
```

```python
#Custom Sorting
#Data → highest priority
#Backend → second
#Web → third
#Alphabetical sorting won't give us that.
#We can craete categorical order
department_order = pd.CategoricalDtype(['Data','Backend','Web'], ordered=True)
#change column type
employees["department"] = employees[
    "department"
].astype(department_order)
#sort
employees.sort_values("department")
```

```python
#Final Challenge
#salary → highest to lowest
employees.sort_values('salary',ascending=False)
```

```python
#department → alphabetical
#salary → highest to lowest
employees.sort_values(['department','salary'],ascending=[True,False])
```

```python
#Find the top 3 performers.
employees.nlargest(3,'performance')
```

```python
employees.loc[:,['name','salary','performance']].sort_values(['performance','salary'],ascending=[False,False])
```

```python
employees.loc[
    (employees['department'].isin(['Data','Backend'])) & 
    (employees['experience'] >= 2),
    ['name','department','experience','salary','performance']
].sort_values(['performance','salary'],ascending=[False,False])
```
