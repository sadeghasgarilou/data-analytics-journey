```python
import pandas as pd

employees = pd.DataFrame({
    "name": [
        "Ali", "Sara", "Reza", "Maryam",
        "Hassan", "Neda", "Omid", "Zahra"
    ],
    "department": [
        "Data", "Web", "Data", "Backend",
        "Web", "Data", "Backend", "Web"
    ],
    "age": [25, 31, 28, 35, 24, 29, 32, 27],
    "salary": [2500, 3200, 2800, 4100, 2300, 3500, 3900, 3000],
    "experience": [2, 5, 3, 8, 1, 6, 7, 4],
    "performance": [82, 91, 76, 95, 68, 88, 93, 85]
})

employees
```

```python
employees['salary'] > 3000
```

```python
employees.loc[employees['salary'] > 3000]
```

```python
#AND, OR, NOT
#Employees with salary >= 3000 AND performance >= 90.
employees.loc[(employees['salary'] > 3000) & (employees['performance'] >= 90)]
```

```python
#Salary >= 4000 OR performance >= 90.
employees[(employees['salary'] >= 4000) | (employees['performance'] >= 90)]
```

```python
#Everyone who is NOT in Data.
employees[~(employees['department']=='Data')]
```

```python
#Put every individual condition inside parentheses.
```

```python
#complex conditions
#Find Data employees who have at least 3 years of experience AND performance >= 80.
employees.loc[(employees['experience'] >= 3) & (employees['department']=='Data') & (employees['performance'] >= 80)]
```

```python
#AND + OR Together
#Find employees who are in Data AND either: performance >= 90 OR salary >= 4000
employees.loc[(employees['department']=='Data') & ((employees['performance'] >= 90) | (employees['salary'] >= 4000))]
```

```python
#Practice 1
#Find employees who are: Web OR Backend AND performance >= 85
employees.loc[(employees['department'].isin(['Data','Backend'])) & (employees['performance'] >= 85)]
```

```python
#between()
employees.loc[employees["age"].between(25,30)]
```

```python
#exclude boundaries
employees.loc[employees['age'].between(25,30,inclusive='neither')]
```

```python
#we can use inclusive="left" and inclusive="right"
employees.loc[employees['age'].between(25,29,inclusive='right')]
```

```python
#Strings + Missing Values
#Find employees whose name contains "a"
employees.loc[employees['name'].str.contains('a',case=False,na=False)]
```

```python
#create missing values
employees.loc[2, "salary"] = None
employees.loc[5, "performance"] = None
```

```python
employees
```

```python
#Find employees with missing salary
employees.loc[employees['salary'].isna()]
```

```python
#Employees whose salary is missing OR performance is below 75.
employees.loc[(employees['salary'].isna()) & (employees['performance'] < 75)]
```

```python
#Show the name, department, salary, and performance of Data or Backend employees who have at least 3 years of experience and performance >= 80.
employees.loc[(employees['department'].isin(['Data','Backend'])) & (employees['experience'] >= 3) & (employees['performance'] >= 80),['name','department','salary','performance']]
```

```python
#final challenge
#Find employees who:
#age between 25 and 32
#AND
#salary >= 3000
employees.loc[(employees['age'].between(25,32,inclusive='neither')) & (employees['salary'] >= 3000)]
```

```python
employees.loc[
    (employees['department'].isin(['Data','Backend'])) &
    (employees['experience'] >= 2) &
    (employees['performance'] >= 80) &
    (employees['salary'].between(2000,4000)),
    ['name','department','salary','performance']
].sort_values(by='performance', ascending=False)
```
