```python
#Load Dataset
import pandas as pd

sales = pd.DataFrame({
    "product": [
        "Laptop", "Mouse", "Keyboard", "Laptop",
        "Monitor", "Mouse", "Keyboard", "Monitor",
        "Laptop", "Mouse"
    ],
    "category": [
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories"
    ],
    "salesperson": [
        "Ali", "Sara", "Ali", "Reza",
        "Sara", "Ali", "Reza", "Sara",
        "Reza", "Sara"
    ],
    "quantity": [2, 10, 5, 1, 3, 15, 8, 2, 2, 12],
    "price": [1000, 25, 50, 1000, 400, 25, 50, 400, 1000, 25]
})

sales["revenue"] = sales["quantity"] * sales["price"]

sales
```

```python
#Filter + Sort
#Example A: Electronics sales, highest revenue first
sales.loc[sales['category']=='Electronics'].sort_values(by='revenue',ascending=False)
```

```python
#Sara’s sales, highest quantity first
sales.loc[sales['salesperson']=='Sara'].sort_values(by='quantity',ascending=False)
```

```python
#Electronics sold by Sara, highest revenue first
sales.loc[(sales['category']=='Electronics') & (sales['salesperson'] == 'Sara')].sort_values(by='revenue',ascending=False)
```

```python
#Filter + Aggregation
#Total revenue from electronics
sales.loc[sales['category'] == 'Electronics','revenue'].sum()
```

```python
#Average revenue for Sara’s sales
sales.loc[sales['salesperson'] == 'Sara','revenue'].mean()
```

```python
#Total quantity for Mouse sales
sales.loc[sales['product'] == 'Mouse','quantity'].sum()
```

```python
#Several summaries for electronics
sales.loc[sales['category'] == 'Electronics','revenue'].agg(['sum','min','max'])
```

```python
#Combine all three skills
#For each salesperson, show their electronics sales, sorted by revenue, and calculate the total revenue from those sales.
sales.loc[sales['category'] == 'Electronics'].groupby('salesperson').sum().sort_values(by='revenue',ascending=False)
```

```python
#Filter Mouse sales and sort them by quantity, highest first.
sales.loc[sales['product'] == 'Mouse'].sort_values(by='quantity', ascending=False)
```

```python
#Find total revenue from sales where quantity is at least 5.
sales.loc[sales['quantity'] >= 5,'revenue'].sum()
```

```python
#Find average revenue for Reza’s sales.
sales.loc[sales['salesperson'] == 'Reza','revenue'].mean()
```

```python
#Filter Accessories sales, sort by price from highest to lowest, and display only product, price, and revenue.
sales.loc[sales['category'] == 'Accessories',['product','price','revenue']].sort_values(by='price',ascending=False)
```

```python
#Calculate the total revenue from Sara’s Electronics sales.
sales.loc[(sales['salesperson'] == 'Sara') & (sales['category'] == 'Electronics'),'revenue'].sum()
```
