```python
#Aggregation → What does the data tell us overall?
```

```python
import pandas as pd
#create our data set
sales = pd.DataFrame({
    "product": [
        "Laptop", "Mouse", "Keyboard", "Laptop",
        "Monitor", "Mouse", "Keyboard", "Monitor",
        "Laptop", "Mouse"
    ],
    "category": [
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories"
    ],
    "salesperson": [
        "Ali", "Sara", "Ali", "Reza",
        "Sara", "Ali", "Reza", "Sara",
        "Reza", "Sara"
    ],
    "quantity": [2, 10, 5, 1, 3, 15, 8, 2, 2, 12],
    "price": [1000, 25, 50, 1000, 400, 25, 50, 400, 1000, 25]
})

sales
```

```python
#calculate revenue
sales['revenue'] = sales['quantity'] * sales['price']
sales
```

```python
#Aggregation means taking many values and producing a summary value.
```

```python
#Basic aggregation functions
#sum = total
sales['revenue'].sum()
```

```python
#mean = average
sales['revenue'].mean()
```

```python
#median = middle value after sorting
sales['revenue'].median()
```

```python
#min = smallest value
sales['revenue'].min()
```

```python
#max = largest value
sales['revenue'].max()
```

```python
#count = number of non missing value
sales['revenue'].count()
```

```python
#nunique = number of unique values
sales['product'].nunique()
```

```python
#unique = return unique values
sales['product'].unique()
```

```python
#practice 1
#Find total revenue.
print(sales['revenue'].sum())
#Find average quantity per transaction.
print(sales['quantity'].mean())
#Find the largest transaction revenue.
print(sales['revenue'].max())
#How many unique products are there?
print(sales['product'].nunique())
#How many unique salespeople?
print(sales['salesperson'].nunique())
```

```python
#Aggregating Multiple Columns
sales[['quantity','price','revenue']].mean()
```

```python
#Different aggregations for different columns
#agg()
sales.agg({
    "quantity":"sum",
    "price":"mean",
    "revenue":"max"
})
```

```python
#Master agg()
#Apply one or more aggregation functions.
#One column, multiple aggregations
sales['revenue'].agg(['max','min','sum','mean'])
```

```python
#Multiple columns, multiple functions
sales[['quantity','revenue']].agg(['sum','mean','median'])
```

```python
#Different functions for different columns
sales.agg({
    "quantity":["sum","median"],
    "revenue":['max','min','mean'],
    "price":["count"]
})
```

```python
#practice 2
#"Give me the total quantity sold, average price, and total revenue."
sales.agg({
    'quantity':['sum'],
    'price':['mean'],
    'revenue':['sum']
})
```

```python
#Aggregation + Filtering
#Total revenue from Laptops
sales.loc[sales['product']=='Laptop','revenue'].sum()
```

```python
#Average revenue for Electronics
sales.loc[sales['category']=='Electronics','revenue'].mean()
```

```python
#Total quantity sold by Sara
sales.loc[sales['salesperson']=='Sara','quantity'].sum()
```

```python
#Total revenue generated by Sara from Electronics:
sales.loc[(sales['category']=='Electronics') & (sales['salesperson']=='Sara'),'revenue'].sum()
```

```python
#Final Challenge
#Calculate:Total revenue
sales['revenue'].sum()
```

```python
#Average revenue per transaction
sales['revenue'].mean()
```

```python
#Total quantity sold for each salesperson
sales.groupby('salesperson')['quantity'].sum()
```

```python
#Total revenue generated by the Electronics category.
sales.loc[sales['category']=='Electronics','revenue'].sum()
```

```python
#The maximum revenue transaction made by Sara.
sales.loc[sales['salesperson']=='Sara','revenue'].max()
```

```python
#Produce one summary containing:
#quantity → total and average
#price    → minimum and maximum
#revenue  → total, average, and maximum
sales.agg({
    "quantity":["sum","mean"],
    "price":["min","max"],
    "revenue":["sum","mean","max"]
})
```
