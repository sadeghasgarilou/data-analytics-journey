```python
import pandas as pd
students = pd.DataFrame({
    "name": ["Ali", "Sara", "Reza", "Maryam", "Hassan", "Neda"],
    "age": [22, 24, 21, 23, 25, 22],
    "math": [85, 92, 70, 96, 60, 88],
    "python": [90, 95, 65, 98, 75, 84],
    "statistics": [78, 89, 72, 94, 64, 90]
})
print(students)
```

```python
#filter based on python scores
students.loc[students['math'] >= 85]
```

```python
#practice 1
#Find students whose python score is greater than 90
students.loc[students['python'] > 90]
```

```python
#Multiple conditions : () & |
students.loc[(students['math'] > 85) & (students['python'] > 90)]
```

```python
students.loc[(students['statistics'] > 90) | (students['math'] > 90)]
```

```python
#Practice 2
#Find students who are: age <= 22 AND python >= 80
students.loc[(students['age'] <= 22) & (students['python'] >= 80)]
```

```python
#Sorting
students.sort_values(by='python')
```

```python
#descending sort
students.sort_values(by='python', ascending=False)
```

```python
#Practice 3
#Sort students by age from youngest to oldest.
students.sort_values(by='age')
```

```python
#Sorting by Multiple Columns
students.sort_values(by=['python','math'],ascending=[False,False])
```

```python
#Filtering + Sorting
#Find students with Python >= 80 and sort them by Python score
students.loc[students['python'] >= 80].sort_values(by='python')
```

```python
#Selecting specific columns 
students.loc[students['python'] >= 80, ['name','python']].sort_values(by='python')
```

```python
#Final mini project
#Task 1
#Find students whose average performance is not directly available yet.
students['average'] = (students['math'] + students['python'] + students['statistics']) / 3
students["average"] = students["average"].round(2)
students
```

```python
#Task 2
#Show students with an average score of 85 or higher.
students.loc[students['average'] >= 85]
```

```python
#Task 3
#Show the top 3 students based on average score.
students_byaverage = students.sort_values(by='average', ascending=False)
students_byaverage.head(3)
```

```python
#Task 4 
#Show only: name, python, average
#for students with: python >= 85 AND average >= 85
#Sort them from highest average to lowest.
new_list = students.loc[(students['python'] >= 85) & (students['average'] >= 85), ['name','python','average']]
new_list.sort_values(by='average', ascending=False)
```
