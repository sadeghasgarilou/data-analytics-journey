```python
#numpy 
import numpy as np
scores = np.array([45,27,21,0])
print(np.mean(scores))
print()
#pandas
import pandas as pd
scores = pd.Series([45,27,21,0])
print(scores)
```

```python
#create series from  a list
import pandas as pd
scores = pd.Series([85,93,64,45,79])
print(scores)
```

```python
#create series with index
scores = pd.Series([85,56,99], index=['Ali','Hasan','Hossein'])
print(scores)
```

```python
#Access to value
print(scores['Ali'])
```

```python
#Change value
scores['Hasan'] = 70
print(scores)
```

```python
#Difference between numpy array and list with Series
#List
scores_list = [85, 92, 78, 95]
print(scores_list[0])
```

```python
#Numpy array
#good for fast computation on numerical array
scores_array = np.array([85,92,78,95])
print(scores_array[0])
print(scores_array.mean())
```

```python
#Pandas Series
scores_series = pd.Series(
    [85,92,78,95],
    index = ['Reza','Soroush','Farshad','Sadegh']
)
#Access with label
print(scores_series['Sadegh'])
print(scores_series.mean())
#Access to index
print(scores_series.index)
#Access with position
print(scores_series.iloc[0])
```

```python
#dtype, shape, and size
print(scores_series.dtype)
print(scores_series.shape)
print(scores_series.size)
```

```python
#Exercise 1
#Create students scores series
import pandas as pd
scores = pd.Series(
    [14,18.5,20,13],
    index = ['Arash', 'Hamid', 'Ali', 'Saber']
)
print(scores)
print(scores['Ali'])
scores['Saber'] = 15.5
print(scores)
print(scores.dtype)
print(scores.size)
print(scores.shape)
```

```python
#Calculate average and min and max score
print(scores.mean())
print(scores.min())
print(scores.max())
```

```python
#filtering
print(scores[scores >= 16])
```

```python
#find name of student with high score
print(scores.idxmax())
```
