```python
#import pandas
import pandas as pd
#define a dataframe
students = pd.DataFrame({
    "name": ["Ali", "Sara", "Reza", "Maryam", "Hassan", "Neda"],
    "age": [22, 24, 21, 23, 25, 22],
    "math": [85, 92, 70, 96, 60, 88],
    "python": [90, 95, 65, 98, 75, 84],
    "statistics": [78, 89, 72, 94, 64, 90]
})
print(students)
```

```python
#select one column
print(students['age'])
```

```python
#select multiple columns
print(students[['name','age']])
```

```python
#Select name, python, and statistics.
print(students[['name','python','statistics']])
```

```python
#Selecting rows with iloc
print(students.iloc[0])
```

```python
#selecting multiple rows
print(students.iloc[0:2])
```

```python
#rows and columns with iloc
#df.iloc[rows, columns]
print(students.iloc[0:2,0:3])
```

```python
#selecting with loc
#iloc → position
#loc → label
#loc becomes especially powerful when we use conditions.
print(students.loc[students['python']>90])
```

```python
#select rows and columns with loc
print(students.loc[students['python']>85,['name','age']])
```

```python
#multiple conditions
print(students.loc[(students['python']>85) & (students['math']>88)])
```
