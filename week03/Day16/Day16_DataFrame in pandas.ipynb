```python
#create DataFrame way 1
students = pd.DataFrame ([
    {"name":"Ali","age":22,"score":85},
    {"name":"Sara","age":24,"score":92},
    {"name":"Reza","age":21,"score":78},
    {"name":"Maryam","age":23,"score":95}
])
print(students)
```

```python
#inspect the DataFrame
#display first rows
print(students.head())
print()
#display only the first two rows
print(students.head(2))
print()
#display the last two rows
print(students.tail(2))
print()
#check the shape
print(students.shape) # 4 rows and 3 columns
print()
#check the column names
print(students.columns)
print()
#check the index
print(students.index)
print()
#check the data types
print(students.dtypes)
print()
#Get general information
print(students.info())
print()
```

```python
#Select columns and rows
print(students['name'])
#this return a Series
print(type(students['name']))
#so DataFrame contains multiple series
```

```python
#Select multiple columns
print(students[['name','age']])
#students["score"]       : Series
#students[["score"]]     : DataFrame
```

```python
#select rows with iloc()
#iloc select data by numerical position
#select first row
print(students.iloc[0])
```

```python
#select first two rows
print(students.iloc[0:2])
```

```python
#select a specefic value
print(students.iloc[1,2])
```

```python
#select rows by loc
#loc select data by index label
print(students.loc[1])
```

```python
#select a specific value
print(students.loc[1,'age'])
```

```python
#filter the DataFrame
#students with scores above 80
high_scores = students[students['score'] > 80]
print(high_scores)
```

```python
#select only the names of high scores students
high_scores_name = students.loc[students['score'] > 80,'name']
print(high_scores_name)
```

```python
#Filter using 2 conditions
#find students who are older than 22 years old and have at least 80 score
high_scores_old_students = students[(students['score'] >= 80) & (students['age']>22)]
print(high_scores_old_students)
# and : &
# or : |
# Always put each condition inside parentheses
```

```python
#Basic Analysis
#calculate averatge
average_score = students['score'].mean()
print(average_score)
```

```python
#Find the highest and lowest score
highest_score = students['score'].max()
lowest_score = students['score'].min()
print(f"Highest score is {highest_score} and lowest score is {lowest_score}")
```

```python
#Find the name of the top student
top_index = students["score"].idxmax()
print(students.loc[top_index, 'name'])
```

```python
#Get Descriptive Statistics
print(students.describe())
```

```python
#Practical Exercise
students = pd.DataFrame({
    "name": ["Ali", "Sara", "Reza", "Maryam", "Hassan", "Neda"],
    "age": [22, 24, 21, 23, 25, 22],
    "math": [85, 92, 70, 96, 60, 88],
    "python": [90, 95, 65, 98, 75, 84],
    "statistics": [78, 89, 72, 94, 64, 90]
})
```

```python
#inspect the data
print(students.shape)
print()
print(students.dtypes)
print()
print(students.size)
print()
print(students.info())
print()
print(students.head())
print()
print(students.columns)
print()
print(students.index)
```

```python
#Display only python columns
print(students['python'])
```

```python
#Find students whose Python score is at least 80.
print(students[students['python'] >= 80])
```

```python
#Calculate the average Math score
print(students['math'].mean())
```

```python
#Find the student with the highest Statistics score.
highest_statistics = students['statistics'].idxmax()
print(students.loc[highest_statistics, 'name' ])
```

```python
#create an average columns
students['average'] = (students['math'] + students['python'] + students['statistics']) / 3
print(students)
```

```python
#sort students
sorted_students = students.sort_values("average", ascending=False)
print(sorted_students)
```
