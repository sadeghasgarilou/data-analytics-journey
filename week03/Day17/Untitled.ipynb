```python
#Review Selecting and filtering
print(students[['name','python']])
```

```python
#select the first row
print(students.iloc[0])
```

```python
#Filter students with python score of at leas 80
print(students.loc[students['python'] >= 80])
```

```python
#Use two conditions
print(students.loc[(students['python'] > 85) & (students['math'] > 85)])
```

```python
#Practical mini project
#create an average columns 
students['average'] = (students['math'] + students['python'] + students['statistics']) / 3
print(students)
```

```python
#round the average
students['average'] = students['average'].round(2)
print(students)
```

```python
#find the top students
top_student_index = students['average'].idxmax()
print(students.loc[top_student_index])
```

```python
#Find Students who passed
passed = students.loc[students['average'] > 70] 
print(passed)
```

```python
#sort students
ranking = students.sort_values('average',ascending=False)
print(ranking)
```

```python
#save result to a new csv file 
ranking.to_csv('sorted_students_with_average.csv',index=False)
```
